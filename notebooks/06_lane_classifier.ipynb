# 06 — Lane & Maneuver Classification

Classify each moment as `straight`, `lane_change`, `turn`, or `sharp_turn`
using heuristic rules on `GYR_Y_DEG` magnitude and duration.

```python
from pathlib import Path
import plotly.express as px

from scooter_imu.data.loader import load_raw
from scooter_imu.data.preprocessor import preprocess
from scooter_imu.algorithms.lane_classifier import LaneClassifier

RAW_PATH = Path("../data/raw/ScooterIMUData.csv")
```

```python
df = preprocess(load_raw(RAW_PATH))
result = LaneClassifier().predict(df)
print(result["maneuver"].value_counts())
```

```python
colour_map = {
    "straight": "lightgray",
    "lane_change": "gold",
    "turn": "royalblue",
    "sharp_turn": "red",
}
fig = px.scatter(result, x="timestamp", y="gyr_y", color="maneuver",
                 color_discrete_map=colour_map,
                 title="Maneuver Classification Timeline",
                 labels={"gyr_y": "GYR_Y (deg/s)"})
fig.show()
```

```python
counts = result["maneuver"].value_counts().reset_index()
counts.columns = ["maneuver", "count"]
fig = px.pie(counts, names="maneuver", values="count",
             color="maneuver", color_discrete_map=colour_map,
             title="Maneuver Distribution")
fig.show()
```
